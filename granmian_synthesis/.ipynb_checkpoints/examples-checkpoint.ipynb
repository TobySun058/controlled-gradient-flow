```python
system = Pendulum(
    g=9.81,  # gravitational constant
    l=1.0,   # lenght
    m=1.0,   # mass
    lam=2.0 # damping
)
x_init = jnp.array([0.0, jnp.pi/2])
x_target = jnp.array([jnp.pi, 0.0])
t_start = 0.0
t_target = 0.75
ode_solver_wrapper = make_solver_wrapper(
    solver=dfx.Dopri8(),
    stepsize_controller=dfx.PIDController(rtol=1e-12, atol=1e-12),
    max_steps=50000,
    dt0=None

)

obj = setup_control_synthesis(
    system=system,
    init_state=x_init,
    target_state=x_target,
    t0=t_start,
    t1=t_target,
    ODE_solver_wrapper=ode_solver_wrapper,
    seed_control="fl_min_energy", #None,
    tol_x=1e-11,
    tol_u=1e-15,
    forward_mode=False,
    integration_method="simpsons",
    integration_samples=5000,
    max_iterations=20,
    min_iterations=10,
    linear_solver_method="ls",
    linear_solver_eps=0.0,
    record_samples=5000,
    interpolation_samples=None #50000
)

```

```python

alpha = [3.75, 1.5]
w = [[1.25, -0.5], [0.25, 1.0]]
b = [[1.0, 0.0], [0.0, 1.0]]  
#b = [[-0.5], [1.0]]
c = [0.0, 0.0]

system = HopfieldRNN(
    D=jnp.array(alpha),
    W=jnp.array(w),
    B=jnp.array(b),
    C=jnp.array(c),
    psi=jax.nn.sigmoid,  #or jnp.tanh,jnp.nn.softplus,jnp.sigmoid
)

x_init = jnp.array([5.0, 7.0])
x_target = jnp.array([1.0, -2.0])
t_start = 0.0
t_target = 2.0

ode_solver_wrapper = make_solver_wrapper(
    solver=dfx.Dopri8(),
    stepsize_controller=dfx.PIDController(rtol=1e-12, atol=1e-12),
    max_steps=5000,
    dt0=None

)

obj = setup_control_synthesis(
    system=system,
    init_state=x_init,
    target_state=x_target,
    t0=t_start,
    t1=t_target,
    ODE_solver_wrapper=ode_solver_wrapper,
    seed_control="feedback",  #"feedback",
    tol_x=1e-10,
    tol_u=1e-15,
    forward_mode=False,
    integration_method="simpsons",
    integration_samples=5000,
    max_iterations=15,
    min_iterations=12,
    linear_solver_method="ls",
    linear_solver_eps=0.0,
    record_samples=5000,
    interpolation_samples=None #50000
)
```

```python
system = load_mindy_model_v2(
    "MINDy100.mat"
)[0]
ode_solver_wrapper = make_solver_wrapper(
    solver=dfx.Dopri8(),
    stepsize_controller=dfx.PIDController(rtol=1e-10, atol=1e-10),
    max_steps=5000,
    dt0=None

)
t_start = 0.0
t_target = 2.0

key = jax.random.PRNGKey(0)
key1, key2 = jax.random.split(key)
x_init = jax.random.normal(key1, system.state_dim)
x_target = jax.random.uniform(key2, system.state_dim) * 2 + 1
obj = setup_control_synthesis(
    system=system,
    init_state=x_init,
    target_state=x_target,
    t0=t_start,
    t1=t_target,
    ODE_solver_wrapper=ode_solver_wrapper,
    seed_control="feedback",  #"feedback",
    tol_x=1e-10,
    tol_u=1e-15,
    forward_mode=False,
    integration_method="simpsons",
    integration_samples=500,
    max_iterations=10,
    min_iterations=10,
    linear_solver_method="ls",
    linear_solver_eps=0.0,
    record_samples=5000,
    interpolation_samples=None #1000
)
```

```python
obj, ag_logs, ag_history_states, ag_P_next = solve_almost_gramian_synthesis(
    control_obj=obj
)
```

```python
obj, g_logs, g_history_states, g_M_next = solve_gramian_synthesis(
    control_obj=obj
)
```
